```python
"""
Efficient Piecewise Training of Deep Structured Models for Semantic Segmentation
Based on Lin et al. CVPR 2016
Adapted for Hugging Face jagennath-hari/nyuv2 dataset

Key components:
1. FeatMap-Net: Multi-scale VGG16 with spatial pyramid pooling
2. Unary-Net: CNN potentials for per-pixel class scores
3. Pairwise-Net: CNN potentials for label pair compatibility
4. Piecewise Training: Avoids expensive CRF inference during training
5. Mean-Field Inference: Approximate inference at test time
"""

import os
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from torchvision import models, transforms
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt
from tqdm import tqdm
from datasets import load_dataset

os.environ["CUDA_LAUNCH_BLOCKING"] = "1"

# ============================================================================
# Data Augmentation (as per paper)
# ============================================================================
class SegmentationAugmentation:
    """
    Data augmentation for semantic segmentation following the paper.
    Includes: random scaling, random crop, random horizontal flip, color jitter
    """
    def __init__(self, base_size=520, crop_size=480, scale_range=(0.5, 2.0)):
        self.base_size = base_size
        self.crop_size = crop_size
        self.scale_range = scale_range
        
        # Color augmentation
        self.color_jitter = transforms.ColorJitter(
            brightness=0.4,
            contrast=0.4,
            saturation=0.4,
            hue=0.1
        )
    
    def __call__(self, image, label):
        """
        Apply augmentation to image and label pair.
        image: PIL Image
        label: PIL Image (grayscale with class indices)
        """
        # Random scaling
        scale = np.random.uniform(self.scale_range[0], self.scale_range[1])
        target_size = int(self.base_size * scale)
        
        image = image.resize((target_size, target_size), Image.BILINEAR)
        label = label.resize((target_size, target_size), Image.NEAREST)
        
        # Random crop
        w, h = image.size
        if w > self.crop_size and h > self.crop_size:
            x1 = np.random.randint(0, w - self.crop_size)
            y1 = np.random.randint(0, h - self.crop_size)
            image = image.crop((x1, y1, x1 + self.crop_size, y1 + self.crop_size))
            label = label.crop((x1, y1, x1 + self.crop_size, y1 + self.crop_size))
        else:
            # Pad if too small
            image = image.resize((self.crop_size, self.crop_size), Image.BILINEAR)
            label = label.resize((self.crop_size, self.crop_size), Image.NEAREST)
        
        # Random horizontal flip
        if np.random.random() > 0.5:
            image = image.transpose(Image.FLIP_LEFT_RIGHT)
            label = label.transpose(Image.FLIP_LEFT_RIGHT)
        
        # Color jittering (only on image)
        if np.random.random() > 0.5:
            image = self.color_jitter(image)
        
        return image, label


class TestAugmentation:
    """Simple resize for test time."""
    def __init__(self, size=(480, 640)):
        self.size = size
    
    def __call__(self, image, label):
        image = image.resize(self.size, Image.BILINEAR)
        label = label.resize(self.size, Image.NEAREST)
        return image, label


# ============================================================================
# Dataset for Semantic Segmentation (NYUv2 with 40 classes from HuggingFace)
# ============================================================================
class NYUv2HFDataset(Dataset):
    """NYUv2 dataset from HuggingFace for semantic segmentation (40 classes)."""
    def __init__(self, split='train', transform=None, augmentation=None, max_images=None):
        self.transform = transform
        self.augmentation = augmentation
        self.num_classes = 40
        self.split = split
        
        # Load dataset from HuggingFace
        print(f"Loading NYUv2 dataset from HuggingFace ({split} split)...")
        self.dataset = load_dataset("jagennath-hari/nyuv2", split=split)
        
        if max_images:
            self.dataset = self.dataset.select(range(min(max_images, len(self.dataset))))
        
        print(f"Loaded {len(self.dataset)} images")
        
        # CRITICAL: NYUv2 uses 16-bit encoded labels, need to create mapping
        print("Building label mapping from 16-bit to class indices...")
        self._build_label_mapping()
        
    def _build_label_mapping(self):
        """Build mapping from 16-bit encoded labels to 0-39 class indices."""
        # Collect all unique labels from dataset
        all_labels = set()
        print("Scanning dataset for unique labels...")
        for i in range(min(100, len(self.dataset))):  # Sample first 100
            semantic = np.array(self.dataset[i]["semantic"])
            all_labels.update(np.unique(semantic).tolist())
        
        # Sort and create mapping (excluding any existing 255 or very high values)
        unique_labels = sorted([l for l in all_labels if l < 65535])
        
        print(f"Found {len(unique_labels)} unique labels in dataset")
        print(f"Label range: [{min(unique_labels)}, {max(unique_labels)}]")
        
        # Create mapping: original_label -> class_index (0-39)
        self.label_mapping = {}
        for idx, original_label in enumerate(unique_labels[:self.num_classes]):
            self.label_mapping[original_label] = idx
        
        # Special handling for 65535 (often used as void/unknown)
        if 65535 in all_labels:
            self.label_mapping[65535] = 255  # Map to ignore_index
        
        print(f"Mapped {len(self.label_mapping)} labels to {self.num_classes} classes")
        print(f"Sample mappings: {list(self.label_mapping.items())[:10]}")

    def _remap_labels(self, label_array):
        """Remap 16-bit encoded labels to 0-39 class indices."""
        # Create output array
        remapped = np.full_like(label_array, 255, dtype=np.int64)
        
        # Remap known labels
        for original, mapped in self.label_mapping.items():
            mask = label_array == original
            remapped[mask] = mapped
        
        return remapped

    def __len__(self):
        return len(self.dataset)

    def __getitem__(self, idx):
        sample = self.dataset[idx]
        
        # Get RGB image and semantic segmentation
        rgb = sample["rgb"]  # PIL Image
        semantic = sample["semantic"]  # PIL Image
        
        # Remap labels from 16-bit to 0-39
        semantic_array = np.array(semantic)
        semantic_remapped = self._remap_labels(semantic_array)
        semantic_pil = Image.fromarray(semantic_remapped.astype(np.uint8))
        
        # Apply augmentation if training
        if self.augmentation is not None:
            rgb, semantic_pil = self.augmentation(rgb, semantic_pil)
        
        # Convert to tensors
        if self.transform:
            rgb = self.transform(rgb)
        else:
            rgb = transforms.ToTensor()(rgb)
        
        # Label to tensor
        label = torch.from_numpy(np.array(semantic_pil)).long()
        
        return rgb.float(), label


# ============================================================================
# FeatMap-Net: Multi-scale VGG16 + Spatial Pyramid Pooling
# ============================================================================
class SpatialPyramidPooling(nn.Module):
    """Sliding spatial pyramid pooling for multi-scale context."""
    def __init__(self, pool_sizes=[1, 2, 4]):
        super().__init__()
        self.pool_sizes = pool_sizes
    
    def forward(self, x):
        B, C, H, W = x.shape
        pooled = [x]
        for ps in self.pool_sizes:
            pooled_feat = F.adaptive_max_pool2d(x, (ps, ps))
            upsampled = F.interpolate(pooled_feat, size=(H, W), mode='bilinear', align_corners=False)
            pooled.append(upsampled)
        return torch.cat(pooled, dim=1)


class FeatMapNet(nn.Module):
    """
    Multi-scale feature extraction network based on VGG16.
    Processes image at multiple scales and applies spatial pyramid pooling.
    """
    def __init__(self, scales=[1.0, 0.5, 0.25], pool_sizes=[1, 2, 4]):
        super().__init__()
        self.scales = scales
        
        # Load pretrained VGG16 features (blocks 1-5)
        vgg = models.vgg16(weights=models.VGG16_Weights.DEFAULT)
        features = list(vgg.features.children())
        
        # Split into blocks for multi-scale processing
        self.block1 = nn.Sequential(*features[0:5])    # -> 64 channels
        self.block2 = nn.Sequential(*features[5:10])   # -> 128 channels
        self.block3 = nn.Sequential(*features[10:17])  # -> 256 channels
        self.block4 = nn.Sequential(*features[17:24])  # -> 512 channels
        self.block5 = nn.Sequential(*features[24:31])  # -> 512 channels
        
        # Block 6: Additional conv block
        self.block6 = nn.Sequential(
            nn.Conv2d(512, 512, 3, padding=1),
            nn.ReLU(inplace=True),
            nn.Conv2d(512, 512, 3, padding=1),
            nn.ReLU(inplace=True),
        )
        
        # Spatial pyramid pooling
        self.spp = SpatialPyramidPooling(pool_sizes)
        
        # Feature dimension after SPP
        spp_mult = 1 + len(pool_sizes)
        self.out_channels = 512 * spp_mult
        
        # Fusion layer for multi-scale features
        self.fusion = nn.Conv2d(self.out_channels * len(scales), 512, 1)
    
    def forward_single_scale(self, x):
        """Process single scale through VGG blocks."""
        x = self.block1(x)
        x = self.block2(x)
        x = self.block3(x)
        x = self.block4(x)
        x = self.block5(x)
        x = self.block6(x)
        x = self.spp(x)
        return x
    
    def forward(self, x):
        B, C, H, W = x.shape
        multi_scale_feats = []
        
        # Target size from largest scale
        target_h, target_w = H // 16, W // 16
        
        for scale in self.scales:
            if scale != 1.0:
                scaled_x = F.interpolate(x, scale_factor=scale, mode='bilinear', align_corners=False)
            else:
                scaled_x = x
            
            feat = self.forward_single_scale(scaled_x)
            
            if feat.shape[2:] != (target_h, target_w):
                feat = F.interpolate(feat, size=(target_h, target_w), mode='bilinear', align_corners=False)
            
            multi_scale_feats.append(feat)
        
        fused = torch.cat(multi_scale_feats, dim=1)
        out = self.fusion(fused)
        return out


# ============================================================================
# Unary-Net: Per-pixel class scores
# ============================================================================
class UnaryNet(nn.Module):
    """Shallow network for unary potentials (per-node class scores)."""
    def __init__(self, in_channels, num_classes):
        super().__init__()
        self.net = nn.Sequential(
            nn.Conv2d(in_channels, 256, 1),
            nn.ReLU(inplace=True),
            nn.Dropout2d(0.5),
            nn.Conv2d(256, num_classes, 1)
        )
    
    def forward(self, feat_map):
        return self.net(feat_map)


# ============================================================================
# Pairwise-Net: Label pair compatibility scores
# ============================================================================
class PairwiseNet(nn.Module):
    """CNN for pairwise potentials modeling label compatibility."""
    def __init__(self, in_channels, num_classes, relation_type='surrounding'):
        super().__init__()
        self.num_classes = num_classes
        self.relation_type = relation_type
        
        self.edge_net = nn.Sequential(
            nn.Conv2d(in_channels * 2, 256, 1),
            nn.ReLU(inplace=True),
            nn.Dropout2d(0.5),
            nn.Conv2d(256, num_classes * num_classes, 1)
        )
    
    def get_shifted_features(self, feat, direction):
        """Get features shifted in a direction for pairwise computation."""
        B, C, H, W = feat.shape
        if direction == 'right':
            shifted = F.pad(feat[:, :, :, :-1], (1, 0, 0, 0))
        elif direction == 'left':
            shifted = F.pad(feat[:, :, :, 1:], (0, 1, 0, 0))
        elif direction == 'down':
            shifted = F.pad(feat[:, :, :-1, :], (0, 0, 1, 0))
        elif direction == 'up':
            shifted = F.pad(feat[:, :, 1:, :], (0, 0, 0, 1))
        return shifted
    
    def forward(self, feat_map):
        """Compute pairwise potentials for neighboring pixels."""
        directions = ['right', 'left', 'down', 'up']
        pairwise_scores = []
        
        for d in directions:
            shifted = self.get_shifted_features(feat_map, d)
            edge_feat = torch.cat([feat_map, shifted], dim=1)
            scores = self.edge_net(edge_feat)
            pairwise_scores.append(scores)
        
        return torch.stack(pairwise_scores, dim=0).mean(dim=0)


# ============================================================================
# Full CRF Model
# ============================================================================
class DeepCRFModel(nn.Module):
    """Deep CRF model combining FeatMap-Net, Unary-Net, and Pairwise-Net."""
    def __init__(self, num_classes=40, use_pairwise=True):
        super().__init__()
        self.num_classes = num_classes
        self.use_pairwise = use_pairwise
        
        self.featmap_net = FeatMapNet(scales=[1.0, 0.5], pool_sizes=[1, 2])
        self.unary_net = UnaryNet(512, num_classes)
        
        if use_pairwise:
            self.pairwise_net = PairwiseNet(512, num_classes, 'surrounding')
        
        self.compat = nn.Parameter(torch.eye(num_classes) * -1.0)
    
    def forward(self, x, return_features=False):
        """Forward pass returning unary scores."""
        feat = self.featmap_net(x)
        unary = self.unary_net(feat)
        
        if return_features:
            return unary, feat
        return unary
    
    def get_pairwise_scores(self, feat):
        """Get pairwise potential scores."""
        if self.use_pairwise:
            return self.pairwise_net(feat)
        return None


# ============================================================================
# Piecewise Training Losses
# ============================================================================
class PiecewiseUnaryLoss(nn.Module):
    """Piecewise loss for unary potentials with class balancing."""
    def __init__(self, ignore_index=255, class_weights=None):
        super().__init__()
        self.ce = nn.CrossEntropyLoss(ignore_index=ignore_index, weight=class_weights)
    
    def forward(self, unary_scores, labels):
        if unary_scores.shape[2:] != labels.shape[1:]:
            unary_scores = F.interpolate(unary_scores, size=labels.shape[1:], 
                                         mode='bilinear', align_corners=False)
        return self.ce(unary_scores, labels)


class PiecewisePairwiseLoss(nn.Module):
    """Piecewise loss for pairwise potentials."""
    def __init__(self, num_classes, ignore_index=255):
        super().__init__()
        self.num_classes = num_classes
        self.ignore_index = ignore_index
    
    def forward(self, pairwise_scores, labels):
        """
        pairwise_scores: (B, C*C, H, W) - compatibility for each label pair
        labels: (B, H_label, W_label) - ground truth labels
        """
        B, CC, H, W = pairwise_scores.shape
        C = self.num_classes
        
        # Resize labels to match pairwise feature map size
        if labels.shape[1:] != (H, W):
            labels_small = F.interpolate(labels.float().unsqueeze(1), size=(H, W), 
                                        mode='nearest').squeeze(1).long()
        else:
            labels_small = labels
        
        # Get shifted labels for horizontal and vertical neighbors
        # Horizontal: compare with right neighbor
        labels_center = labels_small[:, :, :-1]  # All except last column
        labels_right = labels_small[:, :, 1:]     # All except first column
        
        # Vertical: compare with bottom neighbor  
        labels_top = labels_small[:, :-1, :]      # All except last row
        labels_bottom = labels_small[:, 1:, :]    # All except first row
        
        # Create valid masks (both pixels must be valid, not ignore_index)
        valid_h = (labels_center != self.ignore_index) & (labels_right != self.ignore_index)
        valid_v = (labels_top != self.ignore_index) & (labels_bottom != self.ignore_index)
        
        # Flatten spatial dimensions and select valid pairs
        # Horizontal pairs
        B_h, H_h, W_h = labels_center.shape
        pairwise_h = pairwise_scores[:, :, :, :-1]  # Match spatial size with labels
        pairwise_h_flat = pairwise_h.reshape(B_h, CC, -1).permute(0, 2, 1)  # (B, H*W, C*C)
        
        # Vertical pairs
        B_v, H_v, W_v = labels_top.shape
        pairwise_v = pairwise_scores[:, :, :-1, :]  # Match spatial size with labels
        pairwise_v_flat = pairwise_v.reshape(B_v, CC, -1).permute(0, 2, 1)  # (B, H*W, C*C)
        
        # Compute targets: center_label * C + neighbor_label
        target_h = labels_center * C + labels_right
        target_v = labels_top * C + labels_bottom
        
        # Flatten everything
        pairwise_h_flat = pairwise_h_flat.reshape(-1, CC)
        pairwise_v_flat = pairwise_v_flat.reshape(-1, CC)
        target_h_flat = target_h.reshape(-1)
        target_v_flat = target_v.reshape(-1)
        valid_h_flat = valid_h.reshape(-1)
        valid_v_flat = valid_v.reshape(-1)
        
        # Compute loss only on valid pairs
        loss = 0
        count = 0
        
        if valid_h_flat.sum() > 0:
            loss += F.cross_entropy(pairwise_h_flat[valid_h_flat], target_h_flat[valid_h_flat])
            count += 1
        if valid_v_flat.sum() > 0:
            loss += F.cross_entropy(pairwise_v_flat[valid_v_flat], target_v_flat[valid_v_flat])
            count += 1
        
        return loss / max(count, 1)


# ============================================================================
# Mean-Field Inference
# ============================================================================
class MeanFieldInference(nn.Module):
    """Mean-field approximate inference for CRF."""
    def __init__(self, num_classes, num_iterations=5):
        super().__init__()
        self.num_classes = num_classes
        self.num_iterations = num_iterations
        self.register_buffer('spatial_kernel', self._create_spatial_kernel())
        
    def _create_spatial_kernel(self, size=5):
        """Create Gaussian spatial kernel for message passing."""
        x = torch.arange(size) - size // 2
        kernel = torch.exp(-0.5 * (x.unsqueeze(0)**2 + x.unsqueeze(1)**2))
        kernel = kernel / kernel.sum()
        return kernel.view(1, 1, size, size)
    
    def forward(self, unary, pairwise_compat=None, image=None):
        """Mean-field inference."""
        B, C, H, W = unary.shape
        Q = F.softmax(unary, dim=1)
        
        for _ in range(self.num_iterations):
            Q_padded = F.pad(Q, (2, 2, 2, 2), mode='reflect')
            messages = F.conv2d(Q_padded.view(B*C, 1, H+4, W+4), 
                               self.spatial_kernel).view(B, C, H, W)
            
            if pairwise_compat is not None:
                messages = messages.permute(0, 2, 3, 1)
                messages = torch.matmul(messages, pairwise_compat)
                messages = messages.permute(0, 3, 1, 2)
            
            Q = F.softmax(unary - messages, dim=1)
        
        return Q


# ============================================================================
# Training Functions
# ============================================================================
def train_piecewise(model, loader, optimizer, unary_loss_fn, pairwise_loss_fn, 
                    device, train_pairwise=True):
    """Piecewise training: train unary and pairwise potentials independently."""
    model.train()
    total_unary_loss = 0
    total_pairwise_loss = 0
    batch_losses = []
    
    for imgs, labels in tqdm(loader, desc="Training"):
        imgs, labels = imgs.to(device), labels.to(device)
        optimizer.zero_grad()
        
        unary, feat = model(imgs, return_features=True)
        
        loss_u = unary_loss_fn(unary, labels)
        total_unary_loss += loss_u.item()
        batch_losses.append(loss_u.item())
        
        loss_p = torch.tensor(0.0, device=device)
        if train_pairwise and model.use_pairwise:
            pairwise = model.get_pairwise_scores(feat)
            loss_p = pairwise_loss_fn(pairwise, labels)
            total_pairwise_loss += loss_p.item()
        
        loss = loss_u + 0.5 * loss_p
        
        # Check for suspicious losses
        if loss.item() < 1e-6:
            print(f"\n⚠️  WARNING: Loss is suspiciously low: {loss.item():.8f}")
            print(f"   Unary predictions: min={unary.min().item():.4f}, max={unary.max().item():.4f}")
            print(f"   Label range: min={labels.min().item()}, max={labels.max().item()}")
            print(f"   Valid labels: {(labels != 255).sum().item()} / {labels.numel()}")
        
        loss.backward()
        optimizer.step()
    
    n = len(loader)
    avg_loss = total_unary_loss / n
    
    # Report statistics
    if batch_losses:
        print(f"   Loss stats: min={min(batch_losses):.6f}, max={max(batch_losses):.6f}, "
              f"std={np.std(batch_losses):.6f}")
    
    return avg_loss, total_pairwise_loss / n


def evaluate(model, loader, device, use_crf=False, num_classes=40):
    """Evaluate with optional mean-field CRF inference."""
    model.eval()
    
    if use_crf:
        mf = MeanFieldInference(num_classes, num_iterations=5).to(device)
    
    total_correct = 0
    total_pixels = 0
    per_class_correct = torch.zeros(num_classes)
    per_class_total = torch.zeros(num_classes)
    
    with torch.no_grad():
        for batch_idx, (imgs, labels) in enumerate(tqdm(loader, desc="Evaluating")):
            imgs, labels = imgs.to(device), labels.to(device)
            
            unary = model(imgs)
            
            if unary.shape[2:] != labels.shape[1:]:
                unary = F.interpolate(unary, size=labels.shape[1:], 
                                     mode='bilinear', align_corners=False)
            
            if use_crf:
                probs = mf(unary, model.compat)
                preds = probs.argmax(dim=1)
            else:
                preds = unary.argmax(dim=1)
            
            valid = labels != 255
            total_correct += (preds[valid] == labels[valid]).sum().item()
            total_pixels += valid.sum().item()
            
            # Per-class accuracy
            for c in range(num_classes):
                class_mask = (labels == c) & valid
                if class_mask.sum() > 0:
                    per_class_correct[c] += (preds[class_mask] == c).sum().item()
                    per_class_total[c] += class_mask.sum().item()
            
            # Debug first batch
            if batch_idx == 0:
                print(f"\n   First batch debug:")
                print(f"   - Predictions unique: {torch.unique(preds).cpu().tolist()[:20]}")
                print(f"   - Labels unique: {torch.unique(labels[valid]).cpu().tolist()[:20]}")
                print(f"   - Batch accuracy: {(preds[valid] == labels[valid]).float().mean():.4f}")
    
    acc = total_correct / max(total_pixels, 1)
    
    # Show per-class breakdown
    print(f"\n   Per-class accuracy (top 5 classes by pixel count):")
    valid_classes = per_class_total > 0
    class_acc = per_class_correct / (per_class_total + 1e-8)
    sorted_indices = per_class_total.argsort(descending=True)
    
    for i in range(min(5, valid_classes.sum().item())):
        idx = sorted_indices[i].item()
        if per_class_total[idx] > 0:
            print(f"   - Class {idx}: {class_acc[idx]:.4f} "
                  f"({per_class_total[idx]:.0f} pixels)")
    
    return acc


def compute_iou(model, loader, device, num_classes=40):
    """Compute mean IoU."""
    model.eval()
    intersection = torch.zeros(num_classes)
    union = torch.zeros(num_classes)
    
    with torch.no_grad():
        for imgs, labels in loader:
            imgs, labels = imgs.to(device), labels.to(device)
            unary = model(imgs)
            
            if unary.shape[2:] != labels.shape[1:]:
                unary = F.interpolate(unary, size=labels.shape[1:], 
                                     mode='bilinear', align_corners=False)
            
            preds = unary.argmax(dim=1)
            
            for c in range(num_classes):
                pred_c = (preds == c)
                label_c = (labels == c)
                intersection[c] += (pred_c & label_c).sum().item()
                union[c] += (pred_c | label_c).sum().item()
    
    iou = intersection / (union + 1e-8)
    return iou.mean().item(), iou


# ============================================================================
# Visualization
# ============================================================================
def visualize_segmentation(model, loader, device, num_samples=3):
    """Visualize segmentation predictions."""
    model.eval()
    cmap = plt.cm.get_cmap('tab20', 40)
    
    with torch.no_grad():
        for imgs, labels in loader:
            imgs, labels = imgs.to(device), labels.to(device)
            unary = model(imgs)
            
            if unary.shape[2:] != labels.shape[1:]:
                unary = F.interpolate(unary, size=labels.shape[1:], 
                                     mode='bilinear', align_corners=False)
            preds = unary.argmax(dim=1)
            
            for i in range(min(num_samples, imgs.shape[0])):
                fig, axes = plt.subplots(1, 3, figsize=(15, 5))
                
                # RGB
                img_np = imgs[i].cpu().permute(1, 2, 0).numpy()
                img_np = img_np * np.array([0.229, 0.224, 0.225]) + np.array([0.485, 0.456, 0.406])
                axes[0].imshow(np.clip(img_np, 0, 1))
                axes[0].set_title("Input Image")
                axes[0].axis('off')
                
                # Ground truth
                axes[1].imshow(labels[i].cpu(), cmap=cmap, vmin=0, vmax=39)
                axes[1].set_title("Ground Truth")
                axes[1].axis('off')
                
                 # Prediction
                axes[2].imshow(preds[i].cpu(), cmap=cmap, vmin=0, vmax=39)
                axes[2].set_title("Prediction")
                axes[2].axis('off')
                
                plt.tight_layout()
                plt.show()
            break


def visualize_all_stages(stage1_model, stage2_model, stage3_model, loader, device, num_samples=3):
    """
    Visualize predictions from all three training stages on the same images.
    This allows direct comparison of how the model improves across stages.
    """
    stage1_model.eval()
    stage2_model.eval()
    stage3_model.eval()
    
    cmap = plt.cm.get_cmap('tab20', 40)
    
    print("\n" + "="*70)
    print("VISUALIZING ALL STAGES ON SAME IMAGES")
    print("="*70)
    
    with torch.no_grad():
        for batch_idx, (imgs, labels) in enumerate(loader):
            imgs, labels = imgs.to(device), labels.to(device)
            
            # Get predictions from all three stages
            unary1 = stage1_model(imgs)
            unary2 = stage2_model(imgs)
            unary3 = stage3_model(imgs)
            
            # Resize to match label size
            if unary1.shape[2:] != labels.shape[1:]:
                unary1 = F.interpolate(unary1, size=labels.shape[1:], 
                                      mode='bilinear', align_corners=False)
                unary2 = F.interpolate(unary2, size=labels.shape[1:], 
                                      mode='bilinear', align_corners=False)
                unary3 = F.interpolate(unary3, size=labels.shape[1:], 
                                      mode='bilinear', align_corners=False)
            
            preds1 = unary1.argmax(dim=1)
            preds2 = unary2.argmax(dim=1)
            preds3 = unary3.argmax(dim=1)
            
            # Visualize each sample in the batch - SHOWING ALL 3 IMAGES
            for i in range(min(num_samples, imgs.shape[0])):
                # Denormalize image
                img_np = imgs[i].cpu().permute(1, 2, 0).numpy()
                img_np = img_np * np.array([0.229, 0.224, 0.225]) + np.array([0.485, 0.456, 0.406])
                img_np = np.clip(img_np, 0, 1)
                
                gt_img = labels[i].cpu().numpy()
                pred1_img = preds1[i].cpu().numpy()
                pred2_img = preds2[i].cpu().numpy()
                pred3_img = preds3[i].cpu().numpy()
                
                # Calculate accuracy for this image
                valid = labels[i] != 255
                acc1 = (preds1[i][valid] == labels[i][valid]).float().mean().item()
                acc2 = (preds2[i][valid] == labels[i][valid]).float().mean().item()
                acc3 = (preds3[i][valid] == labels[i][valid]).float().mean().item()
                
                # Create figure with 3 rows x 3 columns
                fig, axes = plt.subplots(3, 3, figsize=(15, 15))
                
                # === STAGE 1: UNARY ONLY ===
                axes[0, 0].imshow(img_np)
                axes[0, 0].set_title("Input Image", fontsize=12, fontweight='bold')
                axes[0, 0].axis('off')
                
                axes[0, 1].imshow(gt_img, cmap=cmap, vmin=0, vmax=39)
                axes[0, 1].set_title("Ground Truth", fontsize=12, fontweight='bold')
                axes[0, 1].axis('off')
                
                axes[0, 2].imshow(pred1_img, cmap=cmap, vmin=0, vmax=39)
                axes[0, 2].set_title(f"Stage 1: Unary Only\nAcc: {acc1*100:.2f}%", 
                                    fontsize=12, fontweight='bold', color='blue')
                axes[0, 2].axis('off')
                
                # === STAGE 2: + PAIRWISE ===
                axes[1, 0].imshow(img_np)
                axes[1, 0].set_title("Input Image", fontsize=12, fontweight='bold')
                axes[1, 0].axis('off')
                
                axes[1, 1].imshow(gt_img, cmap=cmap, vmin=0, vmax=39)
                axes[1, 1].set_title("Ground Truth", fontsize=12, fontweight='bold')
                axes[1, 1].axis('off')
                
                axes[1, 2].imshow(pred2_img, cmap=cmap, vmin=0, vmax=39)
                axes[1, 2].set_title(f"Stage 2: + Pairwise\nAcc: {acc2*100:.2f}%", 
                                    fontsize=12, fontweight='bold', color='green')
                axes[1, 2].axis('off')
                
                # === STAGE 3: FINE-TUNED ===
                axes[2, 0].imshow(img_np)
                axes[2, 0].set_title("Input Image", fontsize=12, fontweight='bold')
                axes[2, 0].axis('off')
                
                axes[2, 1].imshow(gt_img, cmap=cmap, vmin=0, vmax=39)
                axes[2, 1].set_title("Ground Truth", fontsize=12, fontweight='bold')
                axes[2, 1].axis('off')
                
                axes[2, 2].imshow(pred3_img, cmap=cmap, vmin=0, vmax=39)
                axes[2, 2].set_title(f"Stage 3: Fine-tuned\nAcc: {acc3*100:.2f}%", 
                                    fontsize=12, fontweight='bold', color='red')
                axes[2, 2].axis('off')
                
                plt.suptitle(f"Sample {batch_idx * imgs.shape[0] + i + 1} - All Stages Comparison", 
                           fontsize=16, fontweight='bold')
                plt.tight_layout()
                plt.show()
                
                # Print statistics for this image
                print(f"\nSample {batch_idx * imgs.shape[0] + i + 1} Statistics:")
                print(f"  Stage 1 Accuracy: {acc1*100:.2f}%")
                print(f"  Stage 2 Accuracy: {acc2*100:.2f}%")
                print(f"  Stage 3 Accuracy: {acc3*100:.2f}%")
                print(f"  Stage 1→2 Improvement: {(acc2-acc1)*100:+.2f}pp")
                print(f"  Stage 2→3 Improvement: {(acc3-acc2)*100:+.2f}pp")
                print(f"  Overall Improvement: {(acc3-acc1)*100:+.2f}pp")
            
            # Only process first batch
            if batch_idx == 0:
                break
    
    print("\n" + "="*70)

# ============================================================================
# Main Training Pipeline
# ============================================================================
def check_dataset_labels(dataset, num_samples=10):
    """Check label statistics in the dataset."""
    print("\n" + "=" * 60)
    print("Checking Dataset Label Statistics")
    print("=" * 60)
    
    all_labels = []
    class_counts = {}
    total_pixels = 0
    ignored_pixels = 0
    
    for i in range(min(num_samples, len(dataset))):
        _, label = dataset[i]
        unique = torch.unique(label)
        all_labels.extend(unique.tolist())
        
        # Count pixels per class
        for lbl in unique:
            lbl_item = lbl.item()
            count = (label == lbl).sum().item()
            
            if lbl_item == 255:
                ignored_pixels += count
            else:
                class_counts[lbl_item] = class_counts.get(lbl_item, 0) + count
            
            total_pixels += count
        
        valid_count = (label != 255).sum().item()
        print(f"Sample {i}: unique labels = {len(unique)}, "
              f"range = [{unique.min().item()}, {unique.max().item()}], "
              f"shape = {label.shape}, "
              f"valid_pixels = {valid_count}/{label.numel()} ({100*valid_count/label.numel():.1f}%)")
    
    all_labels = set(all_labels)
    print(f"\nTotal unique labels across {num_samples} samples: {len(all_labels)}")
    print(f"Label range: [{min(all_labels)}, {max(all_labels)}]")
    print(f"Labels present: {sorted([l for l in all_labels if l != 255])}")
    
    # Show class distribution
    if class_counts:
        print(f"\nClass distribution (top 10 classes by pixel count):")
        sorted_classes = sorted(class_counts.items(), key=lambda x: x[1], reverse=True)[:10]
        for cls, count in sorted_classes:
            pct = 100 * count / (total_pixels - ignored_pixels)
            print(f"  Class {cls}: {count:,} pixels ({pct:.2f}%)")
    
    valid_pct = 100 * (total_pixels - ignored_pixels) / total_pixels
    print(f"\nOverall: {total_pixels - ignored_pixels:,}/{total_pixels:,} valid pixels ({valid_pct:.1f}%)")
    print(f"Ignored (255): {ignored_pixels:,} pixels ({100*ignored_pixels/total_pixels:.1f}%)")
    print("=" * 60 + "\n")


def compute_class_weights(dataset, num_classes=40, num_samples=100):
    """Compute inverse frequency class weights for balancing."""
    print("\nComputing class weights from dataset...")
    class_counts = torch.zeros(num_classes)
    
    for i in range(min(num_samples, len(dataset))):
        _, label = dataset[i]
        for c in range(num_classes):
            class_counts[c] += (label == c).sum().item()
    
    # Inverse frequency weighting
    total = class_counts.sum()
    class_weights = total / (num_classes * class_counts + 1e-8)
    
    # Normalize and clip to reasonable range
    class_weights = class_weights / class_weights.mean()
    class_weights = torch.clamp(class_weights, 0.1, 10.0)
    
    print(f"Class weight range: [{class_weights.min():.2f}, {class_weights.max():.2f}]")
    print(f"Top 5 weighted classes (rare): {torch.topk(class_weights, 5).values.tolist()}")
    print(f"Bottom 5 weighted classes (common): {torch.topk(class_weights, 5, largest=False).values.tolist()}")
    
    return class_weights


def compare_all_stages(stage1_model, stage2_model, stage3_model, loader, device):
    """Quick comparison of the three stages."""
    models = {
        'Stage 1 (Unary Only)': stage1_model,
        'Stage 2 (+ Pairwise)': stage2_model,
        'Stage 3 (Fine-tuned)': stage3_model
    }
    
    results = {}
    
    print("\n" + "="*70)
    print("STAGE COMPARISON")
    print("="*70)
    print(f"{'Stage':<25} {'Pixel Acc':<15} {'Mean IoU':<15} {'# Classes':<15}")
    print("-"*70)
    
    for stage_name, model in models.items():
        model.eval()
        
        total_correct = 0
        total_pixels = 0
        intersection = torch.zeros(40)
        union = torch.zeros(40)
        predicted_classes = set()
        
        with torch.no_grad():
            for imgs, labels in loader:
                imgs, labels = imgs.to(device), labels.to(device)
                
                unary = model(imgs)
                if unary.shape[2:] != labels.shape[1:]:
                    unary = F.interpolate(unary, size=labels.shape[1:], 
                                         mode='bilinear', align_corners=False)
                preds = unary.argmax(dim=1)
                
                predicted_classes.update(torch.unique(preds).cpu().tolist())
                
                valid = labels != 255
                total_correct += (preds[valid] == labels[valid]).sum().item()
                total_pixels += valid.sum().item()
                
                for c in range(40):
                    pred_c = (preds == c)
                    label_c = (labels == c)
                    intersection[c] += (pred_c & label_c & valid).sum().item()
                    union[c] += ((pred_c | label_c) & valid).sum().item()
        
        pixel_acc = total_correct / max(total_pixels, 1)
        iou = intersection / (union + 1e-8)
        mean_iou = iou[union > 0].mean().item()
        
        results[stage_name] = {
            'pixel_acc': pixel_acc,
            'mean_iou': mean_iou,
            'num_classes': len(predicted_classes)
        }
        
        print(f"{stage_name:<25} {pixel_acc*100:>6.2f}%        {mean_iou*100:>6.2f}%        {len(predicted_classes):>3}")
    
    print("-"*70)
    
    # Show improvements
    s1 = results['Stage 1 (Unary Only)']
    s3 = results['Stage 3 (Fine-tuned)']
    
    print(f"\n📈 Improvements (Stage 1 → Stage 3):")
    print(f"  Pixel Accuracy: {s1['pixel_acc']*100:.2f}% → {s3['pixel_acc']*100:.2f}% "
          f"({(s3['pixel_acc']-s1['pixel_acc'])*100:+.2f}pp)")
    print(f"  Mean IoU: {s1['mean_iou']*100:.2f}% → {s3['mean_iou']*100:.2f}% "
          f"({(s3['mean_iou']-s1['mean_iou'])*100:+.2f}pp)")
    print(f"  Classes Predicted: {s1['num_classes']} → {s3['num_classes']} "
          f"({s3['num_classes']-s1['num_classes']:+})")
    
    return results


def main():
    if torch.cuda.is_available():
        torch.cuda.empty_cache()
    
    device = 'cuda' if torch.cuda.is_available() else 'cpu'
    print(f"Device: {device}")
    
    # Hyperparameters
    NUM_CLASSES = 40
    BATCH_SIZE = 4
    EPOCHS_STAGE1 = 30
    EPOCHS_STAGE2 = 20
    EPOCHS_STAGE3 = 15
    LR_PRETRAINED = 1e-4
    LR_NEW = 1e-3
    
    # Data augmentation (as per paper)
    train_augmentation = SegmentationAugmentation(
        base_size=520,
        crop_size=480,
        scale_range=(0.5, 2.0)  # Random scale from 0.5x to 2x
    )
    test_augmentation = TestAugmentation(size=(480, 640))
    
    # Data transforms (normalization only, augmentation handled separately)
    transform = transforms.Compose([
        transforms.ToTensor(),
        transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
    ])
    
    # Create datasets using HuggingFace
    print("Loading datasets from HuggingFace...")
    train_ds = NYUv2HFDataset(split='train', transform=transform, augmentation=train_augmentation)
    test_ds = NYUv2HFDataset(split='test', transform=transform, augmentation=test_augmentation)
    
    # CRITICAL: Check if train/test are actually different
    print(f"\nDataset sizes - Train: {len(train_ds)}, Test: {len(test_ds)}")
    
    # Compute class weights for balanced training
    class_weights = compute_class_weights(train_ds, NUM_CLASSES, num_samples=100)
    class_weights = class_weights.to(device)
    
    # Check dataset labels before training
    print("\n🔍 TRAINING SET:")
    check_dataset_labels(train_ds, num_samples=5)
    
    print("\n🔍 TEST SET:")
    check_dataset_labels(test_ds, num_samples=5)
    
    print("\n💡 Augmentation enabled:")
    print(f"  - Random scale: 0.5x to 2.0x")
    print(f"  - Random crop: 480x480")
    print(f"  - Random horizontal flip")
    print(f"  - Color jittering (brightness, contrast, saturation, hue)")
    
    train_loader = DataLoader(train_ds, batch_size=BATCH_SIZE, shuffle=True, num_workers=2)
    test_loader = DataLoader(test_ds, batch_size=BATCH_SIZE, shuffle=False, num_workers=2)
    
    print("\n" + "=" * 60)
    print("Stage 1: Training Unary Potentials (FCN baseline)")
    print("=" * 60)
    
    model = DeepCRFModel(num_classes=NUM_CLASSES, use_pairwise=False)
    model.to(device)
    
    pretrained_params = list(model.featmap_net.block1.parameters()) + \
                        list(model.featmap_net.block2.parameters()) + \
                        list(model.featmap_net.block3.parameters()) + \
                        list(model.featmap_net.block4.parameters()) + \
                        list(model.featmap_net.block5.parameters())
    new_params = list(model.featmap_net.block6.parameters()) + \
                 list(model.featmap_net.fusion.parameters()) + \
                 list(model.unary_net.parameters())
    
    optimizer = torch.optim.SGD([
        {'params': pretrained_params, 'lr': LR_PRETRAINED},
        {'params': new_params, 'lr': LR_NEW}
    ], momentum=0.9, weight_decay=5e-4)
    
    unary_loss = PiecewiseUnaryLoss(class_weights=class_weights)
    pairwise_loss = PiecewisePairwiseLoss(NUM_CLASSES)
    
    for epoch in range(EPOCHS_STAGE1):
        loss_u, _ = train_piecewise(model, train_loader, optimizer, 
                                     unary_loss, pairwise_loss, device, 
                                     train_pairwise=False)
        acc = evaluate(model, test_loader, device)
        print(f"[Stage1] Epoch {epoch+1}/{EPOCHS_STAGE1} "
              f"Unary Loss: {loss_u:.4f} Acc: {acc:.4f}")
    
    # Save Stage 1 model for comparison
    import copy
    stage1_model = copy.deepcopy(model)
    print("✅ Stage 1 model saved for comparison")
    
    print("\n" + "=" * 60)
    print("Stage 2: Joint Training with Pairwise Potentials")
    print("=" * 60)
    
    model.use_pairwise = True
    model.pairwise_net = PairwiseNet(512, NUM_CLASSES).to(device)
    
    optimizer.add_param_group({'params': model.pairwise_net.parameters(), 
                               'lr': LR_NEW})
    
    for epoch in range(EPOCHS_STAGE2):
        loss_u, loss_p = train_piecewise(model, train_loader, optimizer,
                                          unary_loss, pairwise_loss, device,
                                          train_pairwise=True)
        acc = evaluate(model, test_loader, device, use_crf=True)
        print(f"[Stage2] Epoch {epoch+1}/{EPOCHS_STAGE2} "
              f"Unary: {loss_u:.4f} Pairwise: {loss_p:.4f} Acc: {acc:.4f}")
    
    # Save Stage 2 model for comparison
    stage2_model = copy.deepcopy(model)
    print("✅ Stage 2 model saved for comparison")
    
    print("\n" + "=" * 60)
    print("Stage 3: Piecewise Training (Freeze Backbone)")
    print("=" * 60)
    
    for param in model.featmap_net.parameters():
        param.requires_grad = False
    
    optimizer = torch.optim.SGD(
        filter(lambda p: p.requires_grad, model.parameters()),
        lr=LR_NEW, momentum=0.9, weight_decay=5e-4
    )
    
    for epoch in range(EPOCHS_STAGE3):
        loss_u, loss_p = train_piecewise(model, train_loader, optimizer,
                                          unary_loss, pairwise_loss, device,
                                          train_pairwise=True)
        acc = evaluate(model, test_loader, device, use_crf=True)
        miou, _ = compute_iou(model, test_loader, device)
        print(f"[Stage3] Epoch {epoch+1}/{EPOCHS_STAGE3} "
              f"Acc: {acc:.4f} mIoU: {miou:.4f}")
    
    # Save Stage 3 model
    stage3_model = model
    print("✅ Stage 3 model saved for comparison")
    
    # Run comprehensive comparison
    print("\n" + "🔍"*30)
    print("RUNNING STAGE COMPARISON ANALYSIS")
    print("🔍"*30)
    
    # Compare all stages
    results = compare_all_stages(stage1_model, stage2_model, stage3_model, 
                                 test_loader, device)
    
    print("\n" + "🎨"*30)
    print("VISUALIZING STAGE COMPARISON ON SAME IMAGES")
    print("🎨"*30)
    visualize_all_stages(stage1_model, stage2_model, stage3_model, test_loader, device, num_samples=3)
    
    print("\n=== Training and Analysis Complete ===")
    return model, (stage1_model, stage2_model, stage3_model), results


if __name__ == "__main__":
    print("Starting Deep CRF Semantic Segmentation Training...")
    print("="*70)
    
    # Run the main training pipeline
    model, (stage1, stage2, stage3), results = main()
    
    print("\n" + "="*70)
    print("✅ TRAINING COMPLETE!")
    print("="*70)
    print("\nFinal Results:")
    for stage_name, metrics in results.items():
        print(f"\n{stage_name}:")
        print(f"  - Pixel Accuracy: {metrics['pixel_acc']*100:.2f}%")
        print(f"  - Mean IoU: {metrics['mean_iou']*100:.2f}%")
        print(f"  - Classes Predicted: {metrics['num_classes']}")
    print("\n" + "="*70)
```
